> **Pourquoi les `.gitkeep` ?**

### 🧪 Exercice 5 — Chat latéral `Ctrl+Alt+I` : générer une fonction complète

**Principe :** le chat latéral permet de décrire une fonction complexe en langage naturel.

**Consigne :**
1. Ouvrir le chat latéral : `Ctrl+Alt+I`
2. Coller ce prompt exact :

```
Écris une fonction Python appelée `calculer_taux_defaut` qui :
- prend en paramètre un DataFrame pandas et un nom de colonne de segmentation
- retourne un DataFrame avec le taux de défaut (loan_status) par segment
- trié par taux décroissant
- inclut le nombre de prêts par segment
- ajoute une docstring complète en français
```

3. Copier le code généré dans la cellule ci-dessous
4. Tester la fonction sur `loan_grade` puis sur `person_home_ownership`


````python
import pandas as pd


def calculer_taux_defaut(df: pd.DataFrame, colonne_segmentation: str) -> pd.DataFrame:
    """
    Calcule le taux de défaut par segment à partir d'un DataFrame.

    Cette fonction agrège la colonne `loan_status` par segment défini par
    `colonne_segmentation` et retourne un DataFrame contenant :
    - le taux de défaut moyen par segment,
    - le nombre total de prêts dans chaque segment,
    - les segments triés par taux de défaut décroissant.

    Paramètres
    ----------
    df : pandas.DataFrame
        DataFrame contenant au moins les colonnes `loan_status` et la colonne
        de segmentation fournie en paramètre.
    colonne_segmentation : str
        Nom de la colonne utilisée pour créer les segments.

    Retour
    ------
    pandas.DataFrame
        DataFrame avec les colonnes `colonne_segmentation`, `taux_defaut` et
        `nb_prets`, trié par `taux_defaut` décroissant.

    Raises
    ------
    KeyError
        Si `loan_status` ou la colonne de segmentation est absente.
    ValueError
        Si le DataFrame est vide.
    """
    if df.empty:
        raise ValueError("Le DataFrame ne doit pas être vide.")

    if "loan_status" not in df.columns:
        raise KeyError("La colonne 'loan_status' est absente du DataFrame.")

    if colonne_segmentation not in df.columns:
        raise KeyError(f"La colonne '{colonne_segmentation}' est absente du DataFrame.")

    resultat = (
        df.groupby(colonne_segmentation, dropna=False)["loan_status"]
        .agg(["mean", "count"])
        .rename(columns={"mean": "taux_defaut", "count": "nb_prets"})
        .sort_values("taux_defaut", ascending=False)
        .reset_index()
    )

    resultat.index.name = colonne_segmentation
    return resultat


# ── Test de la fonction ──────────────────────────────────────────────────────
df = pd.read_csv("C:/Users/HP/Desktop/Project Formation/credit-risk-formation/data/raw/credit_risk_dataset.csv")

# Test 1 : segmentation par grade
print(calculer_taux_defaut(df, "loan_grade").head())

# Test 2 : segmentation par type de logement
print(calculer_taux_defaut(df, "person_home_ownership").head())
````
